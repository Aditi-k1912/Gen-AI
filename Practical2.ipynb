```python
import re
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

tf.keras.utils.set_random_seed(42)

data = [
    ("Great product and works perfectly", 1),
    ("Excellent quality and good delivery", 1),
    ("I love this phone, the battery is great", 1),
    ("Very happy with this good purchase", 1),
    ("Excellent sound and worth the money", 1),
    ("Good value for money and easy to use", 1),
    ("Amazing camera and great performance", 1),
    ("Best laptop, excellent performance", 1),
    ("Perfect fit and great quality", 1),
    ("Good packaging and excellent service", 1),
    ("I love these shoes, they are perfect", 1),
    ("Amazing display and good battery", 1),
    ("Great speaker with good sound", 1),
    ("Excellent product and I am happy", 1),
    ("Nice design and good quality", 1),
    ("Best purchase and worth the price", 1),
    ("Great delivery and nice packaging", 1),
    ("Good headphones with amazing sound", 1),
    ("Perfect phone with an excellent camera", 1),
    ("I love this laptop, it is amazing", 1),
    ("Very good service and I am happy", 1),
    ("Excellent shoes and perfect fit", 1),
    ("Great value and good performance", 1),
    ("The camera is amazing and the quality is great", 1),
    ("Nice product and worth buying", 1),
    ("Good battery, great display and nice design", 1),
    ("Happy with the excellent quality", 1),
    ("Best speaker with great sound", 1),
    ("Amazing delivery and perfect packaging", 1),
    ("I love the good quality and great price", 1),
    ("Bad product and stopped working", 0),
    ("Poor quality and bad delivery", 0),
    ("I hate this phone, the battery is bad", 0),
    ("Very disappointed with this poor purchase", 0),
    ("Terrible sound and a waste of money", 0),
    ("Bad value for money and hard to use", 0),
    ("Terrible camera and poor performance", 0),
    ("Worst laptop, terrible performance", 0),
    ("Poor fit and bad quality", 0),
    ("Bad packaging and terrible service", 0),
    ("I hate these shoes, they are poor", 0),
    ("Terrible display and bad battery", 0),
    ("Bad speaker with poor sound", 0),
    ("Terrible product and I am disappointed", 0),
    ("Ugly design and poor quality", 0),
    ("Worst purchase and a waste of money", 0),
    ("Bad delivery and poor packaging", 0),
    ("Poor headphones with terrible sound", 0),
    ("Worst phone with a bad camera", 0),
    ("I hate this laptop, it is terrible", 0),
    ("Very bad service and I am disappointed", 0),
    ("Terrible shoes and poor fit", 0),
    ("Poor value and bad performance", 0),
    ("The camera is terrible and the quality is poor", 0),
    ("Bad product and a waste of money", 0),
    ("Bad battery, poor display and ugly design", 0),
    ("Disappointed with the terrible quality", 0),
    ("Worst speaker with bad sound", 0),
    ("Terrible delivery and bad packaging", 0),
    ("I hate the poor quality and bad price", 0),
]

def clean(text):
    return re.sub(r"[^a-z\s]", "", text.lower())

texts = [clean(t) for t, _ in data]
labels = np.array([l for _, l in data])

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts).toarray()

X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42, stratify=labels
)

model = Sequential([
    Input(shape=(X.shape[1],)),
    Dense(16, activation="relu"),
    Dense(8, activation="relu"),
    Dense(1, activation="sigmoid"),
])
model.compile(optimizer="adam",
              loss="binary_crossentropy",
              metrics=["accuracy"])
model.summary()

history = model.fit(X_train, y_train, epochs=30,
                    batch_size=8, verbose=0)
print("Final Training Accuracy:",
      round(history.history["accuracy"][-1], 2))

loss, acc = model.evaluate(X_test, y_test, verbose=0)
print("Test Accuracy:", round(acc, 2))

new_reviews = [
    "Excellent product, I love it",
    "Very bad quality, waste of money",
]
vec = vectorizer.transform([clean(r) for r in new_reviews])
probs = model.predict(vec.toarray(), verbose=0)
for review, p in zip(new_reviews, probs):
    label = "Positive" if p[0] >= 0.5 else "Negative"
    print(review, "->", label, round(float(p[0]), 2))
```
