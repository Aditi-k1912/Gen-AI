```python
import pandas as pd


feedback = [
    "good",
    "bad",
    "excellent",
    "poor",
    "good"
]

df = pd.DataFrame({"Feedback": feedback})

print("Original Data:")
print(df)

```

```python
unique = []

for item in feedback:
    if item not in unique:
        unique.append(item)

print("\nUnique Values:", unique)
```

```python
encoded_data = []

for item in feedback:
    row = []
    for value in unique:
        if item == value:
            row.append(1)
        else:
            row.append(0)
    encoded_data.append(row)
```

```python
encoded_df = pd.DataFrame(encoded_data, columns=unique)

print("\nOne-Hot Encoded Data:")
print(encoded_df)
```
