```python
import numpy as np
import string

# Customer Feedback
corpus = np.array([
    "Good Service!",
    "Bad Product.",
    "Excellent Quality!",
    "Poor Service."
])

print("Original Corpus:")
print(corpus)
```

```python
corpus_clean = []

for sentence in corpus:
    # Convert to lowercase
    sentence = sentence.lower()

    # Remove punctuation
    for p in string.punctuation:
        sentence = sentence.replace(p, "")

    corpus_clean.append(sentence)

print("\nPreprocessed Corpus:")
print(corpus_clean)
```

```python
tokenized_corpus = []

for sentence in corpus_clean:
    words = sentence.split()
    tokenized_corpus.append(words)

print("\nTokenized Corpus:")
print(tokenized_corpus)
```

```python
vocabulary = []

for sentence in tokenized_corpus:
    for word in sentence:
        if word not in vocabulary:
            vocabulary.append(word)

# Sort the vocabulary
vocabulary.sort()

print("\nVocabulary:")
print(vocabulary)
```

```python
one_hot_vectors = []

for sentence in tokenized_corpus:
    for word in sentence:
        vector = [0] * len(vocabulary)

        index = vocabulary.index(word)
        vector[index] = 1

        one_hot_vectors.append((word, vector))

print("\nOne-Hot Encoding:")
for word, vector in one_hot_vectors:
    print(word, ":", vector)
```

```python
#Bag of Words

bow_matrix = []

for sentence in tokenized_corpus:
    row = []

    for vocab_word in vocabulary:
        count = 0

        for word in sentence:
            if word == vocab_word:
                count += 1

        row.append(count)

    bow_matrix.append(row)

print("\nBag of Words Matrix:")

for i in range(len(corpus)):
    print(corpus_clean[i], ":", bow_matrix[i])
```
